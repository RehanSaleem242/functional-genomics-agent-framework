## 1. Import Required Libraries

```python
import sys
sys.path.insert(0, '..')

from src import data_utils, model_utils
import pandas as pd
import numpy as np
from tqdm import tqdm
import torch

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")
```

## 2. Configure NCBI Access

```python
# Set your email for NCBI API access
YOUR_EMAIL = "your.email@example.com"  # Replace with your email
data_utils.set_ncbi_email(YOUR_EMAIL)
```

## 3. Fetch Bacterial Genome Data

Let's fetch a well-characterized bacterial genome (E. coli K-12) from NCBI.

```python
# Example: E. coli K-12 MG1655 complete genome
accession_id = "NC_000913"  # E. coli K-12

# Fetch the genome
fasta_file = data_utils.fetch_bacterial_genome(accession_id, output_dir="../data")
print(f"Genome saved to: {fasta_file}")
```

## 4. Load and Process Sequences

```python
# Read sequences from the FASTA file
sequences = data_utils.read_fasta_sequences(fasta_file)

print(f"Number of sequences: {len(sequences)}")
print(f"\nFirst sequence info:")
for key, value in sequences[0].items():
    if key != 'sequence':
        print(f"  {key}: {value}")
    else:
        print(f"  sequence: {value[:100]}... (length: {len(value)})")
```

## 5. Load Mamba Model

**Note**: This may take a few minutes on first run as it downloads the model.

```python
# Load a mamba model
# You can use different models depending on availability
# model_name = "state-spaces/mamba-130m"

# For testing, we'll try to load a smaller model
# Replace with your preferred mamba model
model_name = "facebook/opt-125m"  # Using OPT as fallback for testing

model, tokenizer, device = model_utils.load_mamba_model(model_name)
```

## 6. Process Sequences for Model

Truncate sequences to fit model input limits.

```python
# Truncate sequences to 512 bp for processing
truncated_sequences = data_utils.truncate_sequences(sequences, max_length=512)

# Show truncation statistics
truncated_count = sum(1 for s in truncated_sequences if s['truncated'])
print(f"Total sequences: {len(truncated_sequences)}")
print(f"Truncated sequences: {truncated_count}")
print(f"Average length: {np.mean([s['length'] for s in truncated_sequences]):.2f} bp")
```
