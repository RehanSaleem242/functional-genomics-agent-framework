# Pangenome Mamba Model Testing on NCBI Bacterial Data
This notebook demonstrates how to test and validate a mamba model for pangenome analysis using real NCBI bacterial data. Model files (`spyogenes_mamba_weights.pth` and `mamba_tokenizer_config.json`) are included in this directory.

```python
# Import required libraries
import os
import random
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from Bio import SeqIO
import torch

# Set random seeds for reproducibility
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)
```

## 1. Data Loading
Load bacterial genome data from the `data/` directory. Ensure GenBank files are available for analysis.

```python
# List GenBank files in the data directory
DATA_DIR = '../data/'
genbank_files = [f for f in os.listdir(DATA_DIR) if f.endswith('.gb') or f.endswith('.gbk')]
print(f"Found {len(genbank_files)} GenBank files:", genbank_files)
```

## 2. Preprocessing
Parse GenBank files and extract relevant genomic features using BioPython.

```python
# Example: Parse GenBank files and extract gene features
records = []
for gb_file in tqdm(genbank_files, desc="Parsing GenBank files"):
    for record in SeqIO.parse(os.path.join(DATA_DIR, gb_file), "genbank"):
        for feature in record.features:
            if feature.type == "gene":
                records.append({
                    "file": gb_file,
                    "gene": feature.qualifiers.get("gene", [None])[0],
                    "location": str(feature.location),
                })
genes_df = pd.DataFrame(records)
genes_df.head()
```

## 4. Analysis & Visualization
Analyze the model outputs and visualize comparative pangenome results.

```python
# Example: Visualize gene counts per file
plt.figure(figsize=(10, 6))
sns.countplot(data=genes_df, x="file", order=genes_df["file"].value_counts().index)
plt.title("Gene Counts per GenBank File")
plt.xlabel("GenBank File")
plt.ylabel("Gene Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
